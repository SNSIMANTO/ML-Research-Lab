```python
# Simple ML Example: Predict Exam Score from Study Hours

from sklearn.linear_model import LinearRegression
import numpy as np

# Training data (Hours studied vs Score)
hours = np.array([1, 2, 3, 4, 5, 6]).reshape(-1, 1)
scores = np.array([40, 50, 60, 65, 70, 80])

# Create model
model = LinearRegression()

# Train model
model.fit(hours, scores)

# Predict for new data
new_hours = np.array([[7]])
predicted_score = model.predict(new_hours)

print("📚 Study Hours:", new_hours[0][0])
print("🎯 Predicted Exam Score:", round(predicted_score[0], 2))
```
